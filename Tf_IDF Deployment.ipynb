```python
#load the models

import joblib
classifier = joblib.load('model.pkl')
vectorizer = joblib.load('tfidf_vectorizer.pkl')
```

```python
#load the dataset
import pandas as pd
dataset = pd.read_csv('news.csv', index_col=None)
```

```python
dataset['text'][8]
```

```python
#vectorize & predict
x_count = vectorizer.transform([dataset['text'][8]])
output = classifier.predict(x_count)
```

```python
print(f'The given news is "{output[0]}"')
```
